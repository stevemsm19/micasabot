# Exploratory Data Analysis (EDA)

We will follow these phases:

1. **Univariate analysis:** distribution of individual features.
2. **Bivariate analysis:** relationship between features.
3. **Correlation matrix and heatmap:** to identify multicollinearity and strong associations.
4. **Joint exploration with PairPlot:** to visualize patterns and separability across multiple features simultaneously.

## 1. Univariate Analysis: Distribution of Characteristics

**Explanation:**

- **plt.subplot(2, 3, i):** Divides the figure into a grid of 2 rows and 3 columns, assigning each graph to a position.
- **sns.histplot(df[characteristic], kde=True, bins=30):** Plots the histogram of each characteristic with a density curve to visualize the distribution.
- **plt.tight_layout():** Optimizes the spacing between subplots to avoid overlap.

```python
# Define the grid layout
n_cols = 3
n_rows = 2

plt.figure(figsize=(30, 20))

for i, feature in enumerate(group_a.columns, 1):
    plt.subplot(n_rows, n_cols, i)
    sns.histplot(group_a[feature], kde=True, bins=30, color="blue", label="Group A")
    plt.title(f"Distribution of {feature} - Group A")
    plt.xlabel(feature)
    plt.ylabel("Frequency")
    plt.legend()


for i, feature in enumerate(group_b.columns, len(group_a.columns) + 1):
    plt.subplot(n_rows, n_cols, i)
    sns.histplot(group_b[feature], kde=True, bins=30, color="red", label="Group B")
    plt.title(f"Distribution of {feature} - Group B")
    plt.xlabel(feature)
    plt.ylabel("Frequency")
    plt.legend()

plt.tight_layout()
plt.show()
```

### Observations

- The ID field does not provide any useful information, so it can be discarded.
- The distribution in group a does not have a clear bell shape and is more dispersed.
- The distribution in group b has a central peak, and the shape of the trend curve more closely resembles a bell. This indicates that most people in this group have a similar number of steps, with fewer people at the extremes of the range.
- The distribution of the BMI feature is very similar in both groups.

## 2. Bivariate Analysis: Relationship between Variables

**Explanation:**

- sns.scatterplot(...):
    - x=df[feature]: Defines the variable on the X-axis.
    - y=df[‘mpg’]: Defines the variable on the Y-axis. 
- These graphs allow us to visually observe whether there is any relationship (linear trend, for example) between characteristics of group A and group B.

```python
# Function to make the scatter plot
def plot_scatter(group, color, group_name):
    plt.figure(figsize=(10, 8))
    sns.scatterplot(x=group["steps"], y=group["bmi"], color=color)
    plt.title(f"Comparison of Steps vs BMI in {group_name}")
    plt.xlabel("Steps")
    plt.ylabel("BMI")
    plt.tight_layout()
    plt.show()
```

```python
plot_scatter(group_a, color="red", group_name="Group A")
```

```python
plot_scatter(group_b, color="blue", group_name="Group B")
```

### Observations

- Both plots are quite similar.
- The plots do not show a linear correlation or any clear pattern between the number of steps ("Steps") and the body mass index ("BMI").

## 3. Correlation Matrix and Heatmap

**Explanation:**

- **df.corr():** Calculates the correlation (Pearson coefficient) between the numerical variables in the DataFrame.
- sns.heatmap(...):
  - **cmap=‘coolwarm’:** Uses a palette that differentiates between high (positive) and low (negative) values.
  - **annot=True:** Places the numerical values in each cell.
  - **fmt=".2f":** Defines the numerical format.
  - **linewidths=0.5:** Sets the thickness of the dividing lines.

```python
matrix_correlation_a = group_a.drop(columns=["ID"]).corr()

plt.figure(figsize=(8, 6))
sns.heatmap(
    matrix_correlation_a, cmap="coolwarm", annot=True, fmt=".2f", linewidths=0.5
)
plt.title("Correlation Matrix between Variables in Group A")
plt.show()
```

```python
matrix_correlation_b = group_b.drop(columns=["ID"]).corr()

plt.figure(figsize=(8, 6))
sns.heatmap(
    matrix_correlation_b, cmap="coolwarm", annot=True, fmt=".2f", linewidths=0.5
)
plt.title("Correlation Matrix between Variables in Group B")
plt.show()
```

### Observations

- Both groups show negative correlations between their variables, but these correlations are generally weak, as they are closer to 0 than to -1 or 1.
This suggests that, while there may be some inverse relationship between the variables, it is not strong enough to imply a clear or significant pattern.

## 5. Joint Exploration: PairPlot

**Explanation:**

- **sns.pairplot(...):** Creates a grid of scatterplots showing the relationship between each pair of selected variables.
  - **diag\_kind='kde':** Shows density plots (KDE) on the diagonal instead of histograms, making it easier to visualize the distribution of each variable.
  - **hue="Group":** Colors the points according to the group, which helps to visually identify patterns and separability between the two groups.
  - **palette="Set1":** Assigns a distinct color palette for clarity.
  - **.assign(Group=A\B):** Temporarily adds a column with the group (A/B) so they can be used in the visualization without changing the original dataset.

```python
group_a_selected = group_a[["steps", "bmi"]].assign(Group="A")
group_b_selected = group_b[["steps", "bmi"]].assign(Group="B")

combined_data = pd.concat([group_a_selected, group_b_selected])

sns.pairplot(combined_data, hue="Group", diag_kind="kde", palette="Set1")

plt.suptitle("Pairplot: Comparison of Steps in Group A and Group B with BMI", y=1.02)
plt.show()
```

### Observations

- With this comparison, the "monkey" shape becomes much clearer. The only difference is that when Steps is on the Y-axis, the shape tilts to the right. However, when Steps is on the X-axis, the shape appears upright.

## Conclusions of the Exploratory Analysis

1. **Data Distribution:**
   - The ID field is irrelevant and should be discarded.
   - In Group A, the distribution of the Steps feature is more dispersed, lacking a clear central tendency. This suggests that the participants in this group have a wide range of step counts, with no particular concentration around a central value.
   - On the other hand, Group B displays a bell-shaped distribution for the Steps feature, indicating that most participants have a similar number of steps, with fewer participants at the extremes. This implies a more concentrated pattern around the mean.
   - The BMI distributions in both groups appear very similar, indicating that body mass index (BMI) values are distributed in much the same way across both groups.
2. **No Clear Correlation:**
   - There is no significant relationship between Steps and BMI in either group. 
   - The plots show no linear correlation or clear pattern, with the data appearing random.

3. **Monkey Shape Observation:**
   - The scatterplots reveal a "monkey silhouette", with no clear trend. When Steps is on the Y-axis, the shape tilts right, while it appears upright when Steps is on the X-axis.
